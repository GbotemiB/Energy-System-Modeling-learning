```python
import pandas as pd
import numpy as np

from linopy import Model
```

```python
# one decision variable

m = Model()
```

```python
# power plant 1; min 50mw, max 300mw, opx; 3Eur/mw
# power plant 2; min 100mw, max 400mw opx; 4eur/mw

# demand; 500mw

# objective; minimize cost

x = m.add_variables(lower=50, upper=300, name="plant 1")
y = m.add_variables(lower=100, upper=400, name="plant 2")
```

```python
m.add_constraints(x + y == 500)
```

```python
m.add_objective(x * 3 + y * 4)
```

```python
m
```

```python
m.solve(solver_name="highs")
```

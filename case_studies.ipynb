```python
import pypsa

wind_network = pypsa.Network()
wind_network.set_snapshots(range(4))
wind_network.add("Bus", "electricity")
wind_network.add("Bus", "demand")
wind_network.add(
    "Generator",
    "wind turbine",
    bus="electricity",
    p_nom=100,
    p_max_pu=[0.2, 0.8, 0.4, 0.1],
    marginal_cost=1,
)
wind_network.add("Load", "demand", bus="demand", p_set=[20, 80, 40, 10])
wind_network.add("Link", "transmission line", bus0="electricity", bus1="demand", p_nom=100)

wind_network.optimize(solver_name="highs")
wind_network.generators_t.p.round(2)
```

```python

bess = pypsa.Network()
bess.set_snapshots(range(4))
bess.add("Bus", "electricity")
bess.add("Bus", "demand")
bess.add(
    "Generator",
    "solar",
    bus="electricity",
    p_nom=100,
    p_max_pu=[0, 1, 0, 0],
    marginal_cost=0,
)
bess.add("Generator", "grid", bus="electricity", p_nom=100, marginal_cost=50)
bess.add("Load", "demand", bus="demand", p_set=[40, 80, 40, 40])
bess.add(
    "StorageUnit",
    "battery",
    bus="electricity",
    p_nom=30,
    max_hours=2,
    efficiency_store=0.9,
    efficiency_dispatch=0.9,
    cyclic_state_of_charge=True,
    marginal_cost=1,
)
bess.add("Link", "transmission line", bus0="electricity", bus1="demand", p_nom=100)
bess.optimize(solver_name="highs")
bess.storage_units_t.state_of_charge.round(2)
```
